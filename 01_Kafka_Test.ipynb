```python
from pyspark.sql.functions import *
import json
```

```python
kafka_connection_json = dbutils.secrets.get(scope = 'financial-Transactions-scope-3',key = 'kafka_connection_details')

kafka_config = json.loads(kafka_connection_json)

bootstrap_server = kafka_config['bootstrap_server']
api_key = kafka_config['api_key']
api_secret = kafka_config['api_secret']
topic = kafka_config['topic']
```

```python
jaas_config=f'kafkashaded.org.apache.kafka.common.security.plain.PlainLoginModule required username="{api_key}" password="{api_secret}";'
```

```python
sample_batch = (spark.read.format('kafka')
                .option("kafka.bootstrap.servers",bootstrap_server)
                .option("subscribe",topic)
                .option("kafka.security.protocol", "SASL_SSL")  #this and the below three lines are for authentication
                .option("kafka.sasl.mechanism", "PLAIN")
                .option("kafka.sasl.jaas.config", jaas_config)
                .option("startingOffsets","earliest") #this is to read the data that is available from the very beginning
                .load()
                )
```

```python
sample_batch.display()
```

```python
parsed_batch = sample_batch.select(
col("key").cast("string"),
col("value").cast("string"),
col("topic"),
col("partition"),
col("offset"),
col("timestamp"),
col("timestampType")
)
```

```python
parsed_batch.display()
```

```python
parsed_batch.write.saveAsTable("financial_transactions_streaming.bronze.transactions_batch_test")
```

```python
streaming_df= (spark.readStream.format('kafka')
                .option("kafka.bootstrap.servers",bootstrap_server)
                .option("subscribe",topic)
                .option("kafka.security.protocol", "SASL_SSL")  #this and the below three lines are for authentication
                .option("kafka.sasl.mechanism", "PLAIN")
                .option("kafka.sasl.jaas.config", jaas_config)
                .option("startingOffsets","earliest") #this is to read the data that is available from the very beginning
                .load()
                )
```

```python
parsed_stream = streaming_df.select(
col("key").cast("string"),
col("value").cast("string"),
col("topic"),
col("partition"),
col("offset"),
col("timestamp"),
col("timestampType")
)
```

```python
parsed_stream.writeStream.format('delta')\
.outputMode('Append')\
.option('checkpointLocation','/Volumes/financial_transactions_streaming/source/transactions/checkpoint/')\
.trigger(availableNow=True)\
.toTable("financial_transactions_streaming.bronze.transactions_stream_test")
```
